# Introdução à Computação

Computação pode ser entendida como o estudo de processos capazes de representar, processar e transformar informações por meio de algoritmos. De forma simplificada, podemos representar um processo computacional como a transformação de **entradas (inputs)** em **saídas (outputs)** por meio de um conjunto de instruções.

Um sistema computacional recebe dados, realiza operações sobre eles seguindo determinadas regras e produz um resultado.

## Software e Hardware

O **software** é o conjunto de programas, instruções e dados que orientam o funcionamento de um computador. Ele pode ser dividido em diferentes categorias. O **software de sistema** é responsável por fornecer os recursos necessários para o funcionamento do computador, incluindo sistemas operacionais, drivers e utilitários. Já o **software aplicativo** é desenvolvido para realizar tarefas específicas para o usuário, como editores de texto, navegadores e sistemas de gerenciamento.

O **hardware**, por outro lado, corresponde aos componentes físicos de um computador. Entre eles estão a unidade central de processamento (CPU), memória, dispositivos de armazenamento e periféricos.

O hardware executa as instruções fornecidas pelo software, permitindo que os processos computacionais sejam realizados.

## O que é um Algoritmo?

Um **algoritmo** é uma sequência finita de passos ou instruções bem definidas utilizada para realizar uma tarefa ou resolver um problema. Um algoritmo deve possuir instruções claras e precisas, de modo que seja possível seguir suas etapas para alcançar um resultado.

Um algoritmo não precisa necessariamente ser eficiente, mas uma solução eficiente é desejável quando existem diferentes formas de resolver um problema.

Por exemplo, para tomar banho, podemos seguir um algoritmo simples:

1. Abrir o chuveiro.
2. Ajustar a temperatura da água.
3. Entrar no chuveiro.
4. Ensaboar o corpo.
5. Enxaguar o corpo.
6. Desligar o chuveiro.
7. Secar-se com uma toalha.

Dessa forma, algoritmos podem ser utilizados tanto para representar tarefas cotidianas quanto para resolver problemas complexos de computação.

## O que é Programar?

**Programar** é o processo de criar um conjunto de instruções que determina como um computador deve realizar uma determinada tarefa.

Essas instruções são escritas utilizando uma **linguagem de programação** e formam o **código-fonte** de um programa.

O código-fonte descreve o que o programa deve fazer e como essas operações devem ser realizadas. A programação envolve conceitos como lógica, estruturas de controle, tipos de dados, funções, estruturas de dados e sintaxe.

Em linguagens compiladas, como C e C++, o código-fonte passa por um processo de **compilação**, no qual é transformado em uma representação que pode posteriormente ser executada pelo computador.

## O que é uma IDE?

Uma **IDE (Integrated Development Environment)**, ou Ambiente de Desenvolvimento Integrado, é um software que reúne diversas ferramentas utilizadas durante o desenvolvimento de programas.

Uma IDE normalmente fornece recursos como editor de código, integração com compiladores, depuradores, gerenciamento de projetos, navegação pelo código e ferramentas de execução.

Exemplos de IDEs incluem **Visual Studio, Code::Blocks, Eclipse e CLion**.

É importante diferenciar uma IDE de um compilador. A IDE é o ambiente utilizado pelo programador para desenvolver o software, enquanto o **compilador** é a ferramenta responsável por traduzir o código-fonte de uma linguagem compilada para uma forma que possa ser utilizada na construção do programa executável.

## O que é uma Variável?

Uma **variável** é um nome associado a uma região de armazenamento utilizada pelo programa para guardar um valor durante sua execução.

Em C, uma variável possui principalmente um **nome**, um **tipo** e um **valor**.

O **nome** é utilizado para identificar a variável no código.

O **tipo** determina que tipo de dado pode ser armazenado e também influencia a quantidade e a forma de armazenamento utilizada na memória. Alguns exemplos são `int`, `float`, `double` e `char`.

O **valor** corresponde ao dado armazenado pela variável naquele momento e pode ser alterado durante a execução do programa.

Uma variável também está associada a um **endereço de memória**, que identifica a região de armazenamento onde seus dados estão localizados.

Por exemplo:

```c
int idade = 25;
```

Nesse caso, `idade` é o nome da variável, `int` é seu tipo e `25` é seu valor inicial.

O conceito de endereço de memória será especialmente importante posteriormente, quando estudarmos **ponteiros**.

## O que é uma Função?

Uma **função** é um bloco de código responsável por realizar uma determinada tarefa. Ela pode receber dados como entrada, executar operações e opcionalmente retornar um resultado.

Uma função possui algumas características importantes:

**Tipo de retorno:** determina o tipo de dado que a função retorna. Quando uma função não retorna um valor, utiliza-se `void`.

**Nome:** é o identificador utilizado para chamar a função.

**Parâmetros:** representam os dados que a função recebe como entrada. Uma função também pode não possuir parâmetros.

**Corpo:** contém as instruções que serão executadas quando a função for chamada.

Por exemplo:

```c
int soma(int a, int b) {
    return a + b;
}
```

Nesse exemplo, a função `soma` recebe dois valores inteiros, realiza a soma e retorna um valor do tipo `int`.

Uma função também pode não receber parâmetros nem retornar um valor:

```c
void mensagem(void) {
    printf("Olá!\n");
}
```

## O que são Bibliotecas e Cabeçalhos?

Uma **biblioteca** é um conjunto de funcionalidades previamente implementadas que pode ser utilizado por outros programas. Bibliotecas permitem reutilizar código existente em vez de implementar novamente funcionalidades comuns.

Em C, é comum utilizar **arquivos de cabeçalho (header files)** para disponibilizar as declarações de funções, tipos, constantes e outras funcionalidades fornecidas por uma biblioteca.

Alguns dos principais cabeçalhos da biblioteca padrão de C são:

`#include <stdio.h>` fornece funcionalidades de entrada e saída, como `printf` e `scanf`.

`#include <stdlib.h>` fornece diversas funções utilitárias, incluindo alocação dinâmica de memória, conversões e controle de processos.

`#include <string.h>` fornece funções para manipulação de strings, como `strlen` e `strcpy`.

`#include <math.h>` fornece funções matemáticas, como `sqrt`, `pow` e `sin`.

Os arquivos de cabeçalho não representam necessariamente toda a implementação da biblioteca. Eles fornecem principalmente as declarações necessárias para que o compilador saiba como utilizar determinadas funcionalidades.

## Origem de C

A linguagem de programação **C** foi desenvolvida no início da década de 1970 por **Dennis Ritchie**, nos Bell Laboratories, nos Estados Unidos.

C surgiu como uma evolução da linguagem **B**, que por sua vez tinha relação com a linguagem **BCPL (Basic Combined Programming Language)**.

Um dos principais objetivos do desenvolvimento de C era fornecer uma linguagem eficiente e flexível para o desenvolvimento de software de sistema, especialmente para o sistema operacional **UNIX**.

C conseguiu combinar recursos de abstração suficientes para tornar o desenvolvimento mais simples do que em Assembly, mantendo um alto grau de controle sobre memória e recursos do computador.

## Linguagens de Alto e Baixo Nível

As linguagens de programação podem ser classificadas de diferentes formas de acordo com o nível de abstração que oferecem em relação ao hardware.

Uma linguagem de **alto nível** fornece abstrações que tornam a escrita e compreensão dos programas mais próximas da forma como os seres humanos descrevem problemas. Essas linguagens normalmente escondem muitos detalhes relacionados ao funcionamento do hardware.

Exemplos incluem Python, Java e C#.

Uma linguagem de **baixo nível** possui uma relação mais próxima com as instruções executadas pelo processador e oferece menos abstrações em relação ao hardware. **Assembly** é um exemplo de linguagem de baixo nível.

A linguagem **C** possui características de alto nível, como funções, estruturas de controle e tipos de dados, mas também oferece um alto grau de controle sobre memória e recursos do sistema.

Por isso, C é frequentemente descrita como uma linguagem de **alto nível com características de baixo nível**. Algumas classificações também utilizam o termo "nível intermediário", embora essa classificação não seja universal.

## Primeiro Programa em C

Após configurar o ambiente de desenvolvimento, podemos escrever o primeiro programa em C. Um exemplo clássico é o programa **"Olá, Mundo!"**, utilizado para demonstrar a estrutura básica de um programa.

```c
#include <stdio.h> // Inclui a biblioteca padrão de entrada e saída
#include <stdlib.h> // Inclui a biblioteca padrão de utilitários
void main() { // Função principal do programa
    printf("Ola, Mundo!\n"); // Exibe a mensagem "Ola, Mundo!" na tela
}
```

## Como um Programa em C é Executado?

O código escrito pelo programador é chamado de **código-fonte**. Em C, esse código passa por diferentes etapas antes de se tornar um programa executável.

De forma simplificada, o processo pode ser representado como:

**Código-fonte → Pré-processamento → Compilação → Código-objeto → Linkedição → Executável → Execução**

O **pré-processador** processa diretivas como `#include` e `#define`.

O **compilador** analisa o código-fonte e o traduz para uma representação de código de máquina ou código-objeto, dependendo da arquitetura e do compilador.

O **linker (linkeditor)** combina o código-objeto produzido pelo compilador com outras partes necessárias do programa, como bibliotecas, formando o arquivo executável.

O **executável** é então carregado pelo sistema operacional e suas instruções são executadas pelo processador.

Compreender esse processo será importante para entender posteriormente conceitos como **compiladores, headers, bibliotecas, memória, linker, arquivos `.c`, arquivos `.h` e arquivos executáveis**.

## Usando acentos e cedilha em C

Para usar acentos e cedilha em C, precisamos incluir a biblioteca `<locale.h>` e configurar a localidade para o idioma desejado. Isso permite que o programa reconheça e exiba corretamente caracteres acentuados e cedilha. 

```c
#include <stdio.h>
#include <stdlib.h>
#include <locale.h>

void main() {
    setlocale(LC_ALL, ".UTF8"); // Configura a localidade para o padrão UTF-8
    printf("Olá, Mundo!\n"); // Exibe a mensagem com acento e cedilha
}
```
Desta forma, o programa exibirá corretamente a mensagem "Olá, Mundo!" com o acento e a cedilha.

# Variáveis em C 

Vamos ver os principais tipos de variáveis utilizados em C, que são fundamentais para armazenar e manipular informações durante a execução de um programa.

## Como escrever na tela 

Para escrever na tela em C, utilizamos a função `printf()`, que é parte da biblioteca `stdio.h`. Essa função permite exibir mensagens e valores na saída padrão, que é geralmente a tela do terminal. Podemos usar o print da seguinte forma:
- Texto simples: `printf("Olá, Mundo!\n");`
- Valor de uma variável: `printf("%d", variavel);` (onde `%d` é um especificador de formato para inteiros)
- Combinação de texto e variáveis: `printf("A idade é: %d\n", idade);`
- Pular linhas: `printf("\n");` ou `printf("Linha 1\nLinha 2\n");`
- Para exibir diferentes tipos de dados, utilizamos especificadores de formato, como `%d` para inteiros, `%f` para números de ponto flutuante e `%s` para strings.

## Tipos de dados 

As variáveis em C podem ser de diferentes tipos, cada um adequado para armazenar diferentes tipos de informações. Alguns dos principais tipos de dados são:
- `int`: para números inteiros (ex: 1, -5, 42)
- `float`: para números de ponto flutuante (ex: 3.14, -0.001, 2.5)
- `double`: para números de ponto flutuante com maior precisão (ex: 3.1415926535)
- `char`: para caracteres individuais (ex: 'a', 'Z', '1')
- `string`: para sequências de caracteres (ex: "Olá, Mundo!", "C é legal!")
- `bool`: para valores booleanos (verdadeiro ou falso, representados como 1 ou 0)

Em C , é importante declarar o tipo de cada variável antes de utilizá-la, garantindo que o compilador saiba como alocar memória e interpretar os valores corretamente. Alguns exemplos de declaração de variáveis em C são:

```c
int idade = 25; // Variável inteira
float altura = 1.75; // Variável de ponto flutuante
double peso = 70.5; // Variável de ponto flutuante com maior precisão
char letra = 'A'; // Variável de caractere
char nome[] = "Alice"; // Variável de string
bool isTrue = 1; // Variável booleana
```

## Concatenação

Para imprimir múltiplas variáveis ou textos em uma única linha, podemos utilizar a função `printf()` com especificadores de formato. Por exemplo:

```c
char nome[] = "Alice";
int idade = 25;
float altura = 1.75;
printf("Nome: %s, Idade: %d, Altura: %.2f\n", nome, idade, altura);
```
O compilador procura os especificadores de formato (`%s`, `%d`, `%.2f`) na string e substitui cada um deles pelos valores correspondentes das variáveis fornecidas como argumentos adicionais.

## Lendo dados do teclado

Para ler dados do teclado em C, utilizamos a função `scanf()`, que também faz parte da biblioteca `stdio.h`. Essa função permite capturar a entrada do usuário e armazená-la em variáveis. A sintaxe básica é:

```c
scanf("formato", &variavel);
```
O compilador identifica o tipo de dado a ser lido com base no especificador de formato fornecido e salva no endereço de memória indicado pela referência (`&`) da variável correspondente. Por exemplo:

```c
int idade;
printf("Digite sua idade: ");
scanf("%d", &idade); // Lê um número inteiro e armazena na variável idade
```

## Operações matemáticas

Como em qualquer linguagem de programação, C permite realizar operações matemáticas básicas, como adição, subtração, multiplicação e divisão. Além disso, é possível utilizar operadores para manipular variáveis e realizar cálculos mais complexos.

Os principais operadores matemáticos em C são:
- Padrões: `+` (adição), `-` (subtração), `*` (multiplicação), `/` (divisão) e `%` (módulo, que retorna o resto da divisão).
- Absoluto: `abs()` (função que retorna o valor absoluto de um número).


Vamos definir algumas variáveis e realizar operações matemáticas com elas:

```c
int a = 10;
int b = 5;

printf("A soma de %d e %d é: %d\n", a, b, a + b); // Adição
printf("A subtração de %d e %d é: %d\n", a, b, a - b); // Subtração
printf("A multiplicação de %d e %d é: %d\n", a, b, a * b); // Multiplicação
printf("A divisão de %d e %d é: %d\n", a, b, a / b); // Divisão
printf("O resto da divisão de %d por %d é: %d\n", a, b, a % b); // Módulo
printf("O valor absoluto de -3 é: %d\n", abs(-3)); // Valor absoluto
```



# Condicionais em C

É muito comum que um programa precise tomar decisões com base nas condições de entrada ou no estado atual do programa. Para isso, utilizamos **estruturas condicionais**, que permitem executar diferentes blocos de código dependendo de certas condições. Para criar uma estrutura condicional em C, utilizamos a palavra-chave `if`, seguida de uma expressão lógica entre parênteses. Se a expressão for verdadeira, o bloco de código dentro das chaves `{}` será executado. Caso contrário, o programa pode seguir para um bloco `else` ou `else if`, se houver.

Um exemplo simples de estrutura condicional: para uma pessoa ser considerada maior de idade, ela precisa ter 18 anos ou mais, caso contrário, ela é considerada menor de idade.

```c
if (idade >= 18) {
    printf("Você é maior de idade.\n");
} else {
    printf("Você é menor de idade.\n");
}
```

## Operadores de comparação

Podemos utilizar operadores lógicos de comparação para avaliar condições em estruturas condicionais. Alguns dos principais operadores de comparação em C são:

| Operador | Descrição |
|----------|-----------|
| ==       | Igual a |
| !=       | Diferente de |
| <        | Menor que |
| >        | Maior que |
| <=       | Menor ou igual a |
| >=       | Maior ou igual a |


Atraves destes operadores, podemos criar condições mais complexas e tomar decisões baseadas em diferentes critérios. Por exemplo, vamos completar a estrutura condicional para verificar se a pessoa é maior ou menor de idade:

```c
#include <stdio.h>
#include <stdlib.h>

void main() {
    int idade;

    printf("Digite sua idade: ");
    scanf("%d", &idade);

    if (idade >= 18) {
        printf("Você é maior de idade.\n");
    } else {
        printf("Você é menor de idade.\n");
    }
}
```

Também é possivel adicionar outras condições específicas utilizando `else if`. Por exemplo, podemos verificar se a pessoa precisa se alistas (idade = 18) ou se é menor de idade (idade < 18):

```c
#include <stdio.h>
#include <stdlib.h>

void main() {
    int idade;

    printf("Digite sua idade: ");
    scanf("%d", &idade);

    if (idade < 17) {
        printf("Você é menor de idade.\n");
    } else if (idade == 18) {
        printf("Você precisa se alistar.\n");
    } else {
        printf("Você é maior de idade.\n");
    }
}
```

Usamos `else if` no lugar de um simples `if` para verificar uma condição adicional, em um bloco de código se fosse um conjunto de `if`, o programa executaria o primeiro `if` verdadeiro e ignoraria os demais, enquanto com `else if`, podemos verificar condições adicionais de forma sequencial.


## Conectivos Lógicos

Existem situações em que precisamos combinar múltiplas condições para tomar uma decisão. Como por exemplo, se condição 1 **E** condição 2 forem verdadeiras, ou se condição 1 **OU** condição 2 forem verdadeiras. Para isso, utilizamos os **conectivos lógicos**.

Existem três conectivos lógicos principais em C:
- `&&` (E lógico): ele é usando quando precisamos que **todas as condições sejam verdadeiras** para que o resultado final seja verdadeiro.
- `||` (OU lógico): ele é usado quando precisamos que **pelo menos uma das condições seja verdadeira** para que o resultado final seja verdadeiro.
- `!` (NÃO lógico): ele é usado para inverter o valor de uma condição, ou seja, se a condição for verdadeira, o resultado será falso, e vice-versa.

Exemplo de uso dos conectivos lógicos:

```c
if (condicao1 && condicao2) {
    // Executa se ambas as condições forem verdadeiras
}

if (condicao1 || condicao2) {
    // Executa se pelo menos uma das condições for verdadeira
}

if (!condicao1) {
    // Executa se a condição for falsa
}

```

## Condicional Switch Case 

Existem situações que sempre será selecionado apenas uma opção entre várias disponíveis semelhante a escolher uma opção de um menu. Para isso, podemos utilizar a estrutura condicional **switch case**, que permite verificar o valor de uma variável e executar diferentes blocos de código dependendo do valor.

A estrutura básica do switch case é a seguinte:

```c
switch (variavel) {
    case valor1:
        // Bloco de código a ser executado se variavel == valor1
        break;
    case valor2:
        // Bloco de código a ser executado se variavel == valor2
        break;
    // Podemos adicionar quantos casos forem necessários
    default:
        // Bloco de código a ser executado se nenhum dos casos anteriores for verdadeiro
}
```
Para cada valor da variável, podemos definir um bloco de código específico a ser executado. O `break` é utilizado para encerrar a execução do switch case após o bloco correspondente ser executado. Caso nenhum dos casos seja verdadeiro, o bloco `default` será executado.

A diferença do switch case para o if else é que o switch verifica somente uma opção (case x) e executa o bloco de código correspondente, enquanto o if else permite utilizar condições mais complexas, como por exemplo, verificar se uma variável é maior ou menor que um determinado valor. O switch case é mais eficiente quando temos várias opções possíveis para uma variável, enquanto o if else é mais flexível para condições mais complexas.


## Gerar números aleatórios

Para gerar números aleatórios em C, é preciso utilizar duas bibliotecas importantes: `<stdlib.h>` e `<time.h>`. A função `rand()` da biblioteca `<stdlib.h>` é utilizada para gerar números aleatórios, enquanto a função `srand()` da biblioteca `<time.h>` é usada para inicializar o gerador de números aleatórios com uma semente baseada no tempo atual (relógio do computador), garantindo que os números gerados sejam diferentes a cada execução do programa.

Sua sintaxe básica é a seguinte:

```c
#include <stdio.h>
#include <stdlib.h>
void main() {
    srand(time(NULL)); // usa o tempo atual como semente para gerar números aleatórios

    int numero_aleatorio = rand() % 10; // Gera um número aleatório entre 0 e 9

    printf("Número aleatório: %d\n", numero_aleatorio);
}
```
A semente gera um número inicial para o gerador de números aleatórios, e a função `rand()` gera um número pseudoaleatório com base nessa semente. O operador `%` é utilizado para limitar o intervalo do número aleatório gerado. Basicamente vamos pegar o resto da divisão de `rand()` por 10, que resulta no intervalo entre 0 e 9. Por exemplo, 11 % 10 = 1, 21 % 10 = 1, 32 % 10 = 2, e assim por diante. Dessa forma, podemos gerar números aleatórios dentro de um intervalo específico. 

Desta forma, ao lugar `rand() % algum_numero` gera valores entre 0 e `(algum_numero - 1)`, por exemplo, para gerar um número aleatório entre 5 e 14, podemos fazer:

```c
int numero_aleatorio = (rand() % 10) + 5; // Gera um número aleatório entre 5 e 14
```

Também podemos simplificar a nomeclarua usando uma variável para `maximo` e `minimo`:

```c
int minimo = 5;
int maximo = 14;
int numero_aleatorio = (rand() % (maximo - minimo + 1)) + minimo; // Gera um número aleatório entre minimo e maximo
```

# Laço de Repetição em C

Podemos utilizar **laços de repetição** para executar um bloco de código várias vezes, enquanto uma determinada condição for verdadeira. Em C, existem três tipos principais de laços de repetição: `for`, `while` e `do while`.


## Laço de Repetição While

O laço `while` é utilizado quando não sabemos o número de iterações que queremos realizar, mas temos uma condição que determina se o laço deve continuar sendo executado.

Ele funciona da seguinte forma:

```c
ENQUANTO (condição) {
    // Bloco de código a ser executado enquanto a condição for verdadeira
}
```
Caso a condição seja verdadeira, o bloco de código dentro do laço será executado. Após a execução do bloco, a condição é verificada novamente. Esse processo se repete até que a condição se torne falsa, momento em que o laço é encerrado e o programa continua com a execução das instruções seguintes.

Exemplo:

```c
int i = 0;
while (i < 5) {
    printf("O valor de i é: %d\n", i);
    i = i + 1; // Incrementa o valor de i em 1
}
``` 
Para imprimir os números de 0 a 4, o laço `while` verifica se `i` é menor que 5. Enquanto essa condição for verdadeira, ele imprime o valor de `i` e incrementa `i` em 1. Quando `i` chega a 5, a condição se torna falsa e o laço é encerrado.

O incremento de `i` é essencial para evitar um loop infinito, garantindo que a condição eventualmente se torne falsa, ele pode ser feito de diferentes formas, como `i++`, `i += 1` ou `i = i + 1`. Além disso, podemos decrementar `i` utilizando `i--`, `i -= 1` ou `i = i - 1`, dependendo da lógica do programa.

```c
int i = 10;

while (i > -1) {
    printf("O valor de i é: %d\n", i);
    i -= 1; // Decrementa o valor de i em 1
}
```

### While infinito

Uma das maiores utilidades do laço `while` é a possibilidade de criar loops infinitos, que continuam executando enquanto uma condição nunca se torna falsa. Isso pode ser útil em situações onde queremos que o programa continue rodando até que uma ação específica seja tomada, como aguardar por uma entrada do usuário ou monitorar um evento.

Por exemplo, um programa que lê laores até que o valor informado seja 10, isso é chamao de **validação de dados**: o programa continua solicitando a entrada do usuário até que ele digite o valor correto.

```c
int valor;
while (valor != 10) {
    printf("Digite um valor (10 para sair): ");
    scanf("%d", &valor);
    if (valor != 10) {
        printf("Valor incorreto! Tente novamente.\n");
    }
}
printf("Valor correto! Continuando...\n");
```

Também é possivel fazer a validação de dados com **mais de um valor**:

```c
int a = 0;
int b = 0;

while (a < 10 || b < 10) {
    printf("Digite um valor para a (menor que 10): ");
    scanf("%d", &a);
    printf("Digite um valor para b (menor que 10): ");
    scanf("%d", &b);
    
    if (a >= 10 || b >= 10) {
        printf("Um dos valores é maior ou igual a 10! Tente novamente.\n");
    }
}
printf("Valores válidos! Continuando...\n");
```

### Do While

Outra forma de criar um laço de repetição é utilizando o **do while**, que garante que o bloco de código seja executado pelo menos uma vez, mesmo que a condição seja falsa na primeira verificação. A sintaxe do `do while` é a seguinte:

```c
do {
    // Bloco de código a ser executado
} while (condição);
```
Desta forma, o bloco de código dentro do `do` será executado primeiro, e somente depois a condição será verificada. Se a condição for verdadeira, o laço continuará executando; caso contrário, ele será encerrado.

Exemplo:

```c
int i = 10;

do {
    printf("O valor de i é: %d\n", i);
    i -= 1; // Decrementa o valor de i em 1
} while (i < 5);
```
Nesse exemplo, o bloco de código dentro do `do` será executado uma vez, imprimindo o valor de `i` como 10. Em seguida, a condição `i < 5` será verificada. Como a condição é falsa, o laço será encerrado e o programa continuará com as instruções seguintes.


## Laço de Repetição For

Uma outra forma de criar um laço de repetição é utilizando o **for**, nele temos a possibilidade de definir uma variável de controle, uma condição de parada e um incremento ou decremento em uma única linha. A sintaxe do `for` é a seguinte:

```c
for (inicialização; condição; incremento) {
    // Bloco de código a ser executado enquanto a condição for verdadeira
}
```

Exemplo, podemos fazer um laço de repetição que roda 10 vezes:

```c
for (int i = 0; i < 10; i++) {
    printf("O valor de i é: %d\n", i);
}
```
Nesse exemplo, a variável `i` é inicializada com o valor 0, a condição `i < 10` é verificada antes de cada iteração, e o incremento `i++` é realizado após cada execução do bloco de código. O laço continuará executando enquanto a condição for verdadeira, ou seja, enquanto `i` for menor que 10. Quando `i atingir 10, a condição se tornará falsa e o laço será encerrado.



# Vetores, strings e matrizes

## Constantes

As constantes são valores que não podem ser alterados durante a execução do programa. Elas são úteis para representar valores fixos, como limites, taxas ou configurações que não devem ser modificadas. A principal vantagem de utilizar constantes é que elas ocupam menos espaço na memória, pois não precisam ser armazenadas como variáveis mutáveis. 

As constantes podem ser definidas utilizando a palavra-chave `const` ou a diretiva `#define`. A diferença entre elas é que `const` cria uma variável constante com um tipo específico, enquanto `#define` cria uma substituição de texto no pré-processamento.

```c
#define PI 3.14159 // Define uma constante chamada PI com o valor 3.14159
const int MAX_TAM = 100; // Define uma constante inteira chamada MAX_TAM com o valor 100
```

Uma boa prática para definir constantes é utilizar letras maiúsculas para o nome da constante, facilitando a identificação de que se trata de um valor fixo no código. Além disso, é importante escolher nomes descritivos para as constantes, tornando o código mais legível e compreensível.


## Vetores

Um **vetor** é uma estrutura de dados que permite armazenar uma coleção de elementos do mesmo tipo em uma única variável. Cada elemento do vetor é identificado por um índice, que indica sua posição dentro do vetor. Em C, os vetores são declarados especificando o tipo dos elementos e o tamanho do vetor entre colchetes. A sua sintaxe básica é a seguinte:

```c
tipo nome_do_vetor[tamanho];

vetor[3] = {valor1, valor2, valor3}; // Inicialização de um vetor com 3 elementos
```

Os vetores podem armazenar apenas um tipo de dado, como inteiros, caracteres ou números de ponto flutuante. O tamanho do vetor deve ser definido no momento da declaração e não pode ser alterado durante a execução do programa.

```c
int numeros[5] = {1, 2, 3, 4, 5}; // Vetor de inteiros com 5 elementos
char letras[4] = {'A', 'B', 'C', 'D'}; // Vetor de caracteres com 4 elementos
float notas[3] = {8.5, 7.0, 9.5}; // Vetor de números de ponto flutuante com 3 elementos
```

Para trabalhar com vetores, podemos usar alguns comandos e funções úteis, como:

| Comando/Função | Descrição |
|----------------|-----------|
| `vetor[i] = valor;` | Atribuição direta a um elemento do vetor |
| `int variavel = vetor[i];` | Atribuição de um elemento do vetor a uma variável |
| `scanf("%d", &vetor[i]);` | Leitura de um valor do teclado e armazenamento em um elemento do vetor |
| `printf("%d", vetor[i]);` | Impressão de um elemento do vetor na tela |


### Printf e scanf com vetores

Podemos **imprimir** os elementos de um vetor utilizando um laço de repetição, como o `for`, para percorrer cada índice do vetor e exibir seus valores. Por exemplo:

```c
#define TAM 5 // Define uma constante para o tamanho do vetor
int numeros[TAM] = {1, 2, 3, 4, 5}; // Vetor de inteiros com 5 elementos

for (int i = 0; i < TAM; i++) {
    printf("Elemento %d: %d\n", i, numeros[i]); // Imprime o índice e o valor do elemento
}
```

Para **ler** valores do teclado e armazená-los em um vetor, podemos utilizar a função `scanf()` dentro de um laço de repetição. Por exemplo:

```c
#define TAM 5 // Define uma constante para o tamanho do vetor
int numeros[TAM]; // Vetor de inteiros com 5 elementos

for (int i = 0; i < TAM; i++) {
    printf("Digite o valor para o elemento %d: ", i);
    scanf("%d", &numeros[i]); // Lê um valor do teclado e armazena no elemento do vetor
}
```

### Exemplos de funcionamento 

Vamos escrever 3 formas direfentes de definir vetores:
1. Sem definir o tamanho do vetor, mas inicializando com valores:
2. Definindo o tamanho do vetor e inicializando com valores:
3. Definindo o tamanho do vetor com constante: 

Depois 3 formas de atribuir valroes novos a um vetor, lembrando de sempre usar o `fflush(stdin)` para limpar o buffer do teclado antes de ler novos valores:

```c
#include <stdio.h>
#include <stdlib.h>
#define TAM 5 // Define uma constante para o tamanho do vetor

int maint(){
    // Definindo Vetores de várias formas: 
    int vetorInt[] = {1, 2, 3, 4, 5}; // Vetor de inteiros sem tamanho definido
    float vatorFloat[3] = {1.1, 2.2, 3.3}; // Vetor de float com tamanho definido
    char vetorChar[TAM] = {'A', 'B'} // Vetor de char com tamanho definido e constante

    // Modificando valor por atribuição 
    vetorInt[0] = 9; // Atribuindo um novo valor ao primeiro elemento do vetor

    // Modificando valor com variável
    float novoValor = 4.4;
    vatorFloat[1] = novoValor; // Atribuindo um novo valor ao segundo elemento do vetor

    //Modificando valor com usuário
    printf("Digite um valor para o terceiro elemento do vetor de char: ");
    fflush(stdin); // Limpa o buffer do teclado
    scanf("%c", &vetorChar[0]); // Lê um valor do teclado

    // Imprimindo os valores dos vetores
    printf("Vetor de inteiros: ");
    for (int i = 0; i < 5; i++) {
        printf("%d ", vetorInt[i]);
    }

    printf("\nVetor de float: ");
    for (int i = 0; i < 3; i++) {
        printf("%.1f ", vatorFloat[i]);
    }

    printf("\nVetor de char: ");
    for (int i = 0; i < TAM; i++) {
        printf("%c ", vetorChar[i]);
    }
    printf("\n");
    return 0;
}
```


## String (Vetores de Char)

Uma **string** é uma sequência de caracteres armazenada em um vetor de `char`. Em C, as strings são representadas como arrays de caracteres terminados com o caractere nulo `\0`, que indica o final da string. A sintaxe básica para declarar uma string é a seguinte:

```c
char nome[20]; // Declara uma string com capacidade para 19 caracteres + '\0
char sobrenome[] = "Silva"; // Declara e inicializa uma string com o valor "Silva"
```

Pela definição da linguagem C, uma string é um vetor de caracteres, e cada caractere ocupa uma posição no vetor. O último caractere da string é sempre o caractere nulo `\0`, que indica o final da string. Por exemplo, a string `"Olá"` é representada internamente como:

```
Índice: 0   1   2   3
Valor:  'O' 'l' 'á' '\0'
```

### printf, puts e scanf com strings

Para imprimir uma string na tela, podemos utilizar a função `printf()` com o especificador de formato `%s`. Por exemplo:

```c
char nome[] = "Alice Nascimento"; // Declara e inicializa uma string com o valor "Alice Nascimento"
printf("Nome: %s\n", nome); // Imprime a string "Alice Nascimento" na tela
```

Também podemos usar `puts()` para imprimir uma string, que é uma função mais simples e direta:

```c
char nome[] = "Alice Nascimento"; // Declara e inicializa uma string com o valor "Alice Nascimento"
puts(nome); // Imprime a string "Alice Nascimento" na tela, adicionando automaticamente uma nova linha ao final
```

Para ler uma string do teclado, podemos utilizar a função `scanf()` com o especificador de formato `%s`. No entanto, é importante lembrar que `scanf()` lê apenas até o primeiro espaço em branco, então se quisermos ler uma string com espaços, devemos usar `fgets()` ou `gets()`.

Observe que é importante limpar o buffer do teclado antes de ler uma string com espaços, utilizando `fflush(stdin)` ou `getchar()` para evitar problemas de leitura.

```c
char nome[50]; // Declara uma string com capacidade para 49 caracteres + '\0
printf("Digite seu nome completo: ");

fflush(stdin); // Limpa o buffer do teclado

fgets(nome, sizeof(nome), stdin); // Lê uma string do teclado com espaços e armazena na variável
printf("Nome digitado: %s\n", nome); // Imprime a string digitada
```

### Exemplos de funcionamento 

Primeiramente vamos definir uma String pré-determinada, depois mostrar como imprimir quando não tem espaços e quando tem espaços. Não esqueça de sempre limpara o buffer do tecladao antes de ler uma string com espaços, utilizando `fflush(stdin).

A função `fgets` é usada para ler strings usando espaços e `puts` para imprimir strings com espaços. A função `scanf` é usada para ler strings sem espaços e `printf` para imprimir strings sem espaços.

```c
#include <stdio.h>
#include <stdlib.h>

int main() {
    // Definindo uma string pré-determinada
    char nome[50] = "Alice Nascimento"; // Declara e inicializa uma string com o valor "Alice Nascimento"
    printf("Nome pré-determinado: %s\n", nome); // Imprime a string pré-determinada

    // Lendo uma string sem espaços
    char sobrenome[50]; // Declara uma string com capacidade para 49 caracteres + '\0
    printf("Digite seu sobrenome (sem espaços): ");
    scanf("%s", sobrenome); // Lê uma string do teclado sem espaços e armazena na variável
    printf("Sobrenome digitado: %s\n", sobrenome); // Imprime a string digitada

    // Lendo uma string com espaços
    char endereco[100]; // Declara uma string com capacidade para 99 caracteres + '\0
    printf("Digite seu endereço (com espaços): ");
    
    fflush(stdin); // Limpa o buffer do teclado
    
    fgets(endereco, sizeof(endereco), stdin); // Lê uma string do teclado com espaços e armazena na variável
    printf("Endereço digitado: %s\n", endereco); // Imprime a string digitada

    return 0;
}
```


## Matrizes 

Uma **matriz** é um vetor multi-dimensional que armazena elementos do mesmo tipo em duas dimensões, ou seja, em linhas e colunas. Em C, as matrizes são representadas como arrays bidimensionais, onde cada elemento é identificado por dois índices: um para a linha e outro para a coluna. A sintaxe básica para declarar uma matriz é a seguinte:

```c
tipo nome_da_matriz[linhas][colunas];
matriz[2][3] = {{valor1, valor2, valor3}, {valor4, valor5, valor6}}; // Inicialização de uma matriz com 2 linhas e 3 colunas
```

Para atribuir valores a uma matriz, podemos utilizar a notação de índices, especificando a linha e a coluna do elemento que desejamos modificar. Por exemplo:

```c
int matriz[2][3]; // Declara uma matriz de inteiros com 2 linhas e 3 colunas
matriz[0][0] = 1; // Atribuindo o valor 1 à posição (0,0) da matriz

scanf("%d", &matriz[1][2]); // Lê um valor do teclado e armazena na posição (1,2) da matriz
```

### printf e scanf com matrizes

Para imprimir os elementos de uma matriz na tela, podemos utilizar um **laço de repetição aninhado**, onde o primeiro laço percorre as linhas e o segundo laço percorre as colunas da matriz. Por exemplo:

```c
int matriz[2][3] = {{1, 2, 3}, {4, 5, 6}}; 

for (int i = 0; i < 2; i++) { // Laço para percorrer as linhas
    for (int j = 0; j < 3; j++) { // Laço para percorrer as colunas
        printf("Elemento [%d][%d]: %d\n", i, j, matriz[i][j]); // Imprime o índice e o valor do elemento
    }
}
```

Para ler valores do teclado e armazená-los em uma matriz, podemos utilizar a função `scanf()` dentro de um laço de repetição aninhado. Por exemplo:

```c
int matriz[2][3]; 

for (int i = 0; i < 2; i++) { // Laço para percorrer as linhas
    for (int j = 0; j < 3; j++) { // Laço para percorrer as colunas
        printf("Digite o valor para o elemento [%d][%d]: ", i, j);
        scanf("%d", &matriz[i][j]); // Lê um valor do teclado e armazena na posição (i,j) da matriz
    }
}
```

### Exemplos de funcionamento

Primeiramente definimos as constantes para o taamnho da matriz, depois definimos a matriz e imprimimos os valores dela. Em seguida, lemos os valores do teclado e imprimimos novamente.

```c
#include <stdio.h>
#define LINHAS 2 // Define uma constante para o número de linhas da matriz
#define COLUNAS 3 // Define uma constante para o número de colunas da matriz

int main() {
    int matriz[LINHAS][COLUNAS] = {{1, 2, 3}, {4, 5, 6}}; // Declara e inicializa uma matriz de inteiros com 2 linhas e 3 colunas

    // Imprimindo os valores da matriz
    printf("Valores da matriz:\n");
    for (int i = 0; i < LINHAS; i++) { // Laço para percorrer as linhas
        for (int j = 0; j < COLUNAS; j++) { // Laço para percorrer as colunas
            printf("%d ", matriz[i][j]); // Imprime o valor do elemento
        }
        printf("\n"); // Pula para a próxima linha após imprimir todos os elementos da linha atual
    }

    // Lendo valores do teclado e armazenando na matriz
    printf("Digite os valores da matriz:\n");
    for (int i = 0; i < LINHAS; i++) { // Laço para percorrer as linhas
        for (int j = 0; j < COLUNAS; j++) { // Laço para percorrer as colunas
            printf("Elemento [%d][%d]: ", i, j);
            scanf("%d", &matriz[i][j]); // Lê um valor do teclado e armazena na posição (i,j) da matriz
        }
    }

    // Imprimindo novamente os valores da matriz após a leitura do teclado
    printf("Valores da matriz após a leitura:\n");
    for (int i = 0; i < LINHAS; i++) { // Laço para percorrer as linhas
        for (int j = 0; j < COLUNAS; j++) { // Laço para percorrer as colunas
            printf("%d ", matriz[i][j]); // Imprime o valor do elemento
        }
        printf("\n"); // Pula para a próxima linha após imprimir todos os elementos da linha atual
    }

    return 0;
}
```
