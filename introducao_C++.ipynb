# Revisão de fundamentos e introdução ao C++

## Introdução
O C++ é uma linguagem de programação de propósito geral que suporta programação procedural, orientada a objetos e genérica. Ele é amplamente utilizado para desenvolvimento de software, jogos, sistemas embarcados e aplicações de alto desempenho.

## Definição de Variáveis (C e C++)
No C++, as variáveis são usadas para armazenar dados que podem ser manipulados durante a execução de um programa. A definição de variáveis envolve especificar o tipo de dado e o nome da variável.

Em C, a definição de uma variável é feita da seguinte forma:

```c
int idade; // Declaração de uma variável inteira chamada idade
double salario; // Declaração de uma variável do tipo double chamada salario
char letra; // Declaração de uma variável do tipo char chamada letra
bool ativo; // Declaração de uma variável do tipo bool chamada ativo
```

Em C++, a definição de uma variável é semelhante, mas com algumas diferenças na sintaxe e nas funcionalidades:

```cpp
int idade; // Declaração de uma variável inteira chamada idade
double salario; // Declaração de uma variável do tipo double chamada salario
char letra; // Declaração de uma variável do tipo char chamada letra   
bool ativo; // Declaração de uma variável do tipo bool chamada ativo
```

### Imprimindo em C++

Imprimindo os valores das variáveis em C++, primeiramente precisamos incluir a biblioteca de entrada e saída padrão, que nos permite utilizar o objeto `std::cout` para exibir informações no console. A seguir, podemos atribuir valores às variáveis e imprimi-los.

```cpp
#include <iostream>
#include <string>

using namespace std;

int main() {
    int idade = 25; // Atribuindo valor à variável idade
    double salario = 3500.50; // Atribuindo valor à variável salario
    char letra = 'A'; // Atribuindo valor à variável letra
    bool ativo = true; // Atribuindo valor à variável ativo
    string nome = "Alice"; // Atribuindo valor à variável nome

    // Imprimindo os valores das variáveis
    cout << "Idade: " << idade;
    cout << "Salário: " << salario;
    cout << "Letra: " << letra;
    cout << "Ativo: " << (ativo ? "Sim" : "Não");
    cout << "Nome: " << nome;

    return 0;
}
```

### Lendo valores do usuário em C++

Para ler valores do usuário em C++, utilizamos o objeto `std::cin`, que permite capturar a entrada do teclado. A seguir, um exemplo de como ler diferentes tipos de dados:

```cpp
#include <iostream>
using namespace std;

int main() {
    int idade;
    double salario;
    char letra;
    bool ativo;
    string nome;

    // Solicitando ao usuário que insira os valores
    cout << "Digite sua idade: ";
    cin >> idade;

    cout << "Digite seu salário: ";
    cin >> salario;

    cout << "Digite uma letra: ";
    cin >> letra;

    cout << "Digite se está ativo (1 para sim, 0 para não): ";
    cin >> ativo;

    cout << "Digite seu nome: ";
    cin >> nome;

    return 0;
}
```

## Vetores e matrizes em C++

Em C++, vetores e matrizes são estruturas de dados que permitem armazenar múltiplos valores do mesmo tipo. Um vetor é uma coleção unidimensional, enquanto uma matriz é uma coleção bidimensional.

### Vetores
Um vetor é declarado especificando o tipo de dado e o tamanho do vetor. A seguir, em C++ declaramos um vetor de forma semelhante ao C:

```cpp
#include <iostream>
using namespace std;

int numeros[3]; // Declaração de um vetor de inteiros com 5 elementos

numeros[0] = 10; // Atribuindo valor ao primeiro elemento do vetor
numeros[1] = 20; // Atribuindo valor ao segundo elemento do vetor
numeros[2] = 30; // Atribuindo valor ao terceiro elemento do vetor

// ou podemos inicializar o vetor diretamente na declaração
int numeros[3] = {10, 20, 30}; // Inicializando o vetor com valores

// Imprimindo os valores do vetor
// Com FOR loop
for (int i = 0; i < 3; i++) {
    printf("Elemento %d: %d\n", i, numeros[i]); // Em C
    cout << "Elemento " << i << ": " << numeros[i]; // Em C++
}

//Com WHILE loop
int i = 0;
while (i < 3) {
    printf("Elemento %d: %d\n", i, numeros[i]); // Em C
    cout << "Elemento " << i << ": " << numeros[i]; // Em C++
    i++;
}
```

### Matrizes

Uma matriz é uma coleção bidimensional de elementos, onde cada elemento é acessado por dois índices: linha e coluna. A declaração de uma matriz em C++ é feita especificando o tipo de dado, o número de linhas e o número de colunas.

```cpp
#include <iostream>
using namespace std;

int main() {
    int matriz[2][3]= { {1, 2, 3}, {4, 5, 6} }; // Declaração e inicialização de uma matriz 2x3

    // Imprimindo os valores da matriz
    for (int i = 0; i < 2; i++) {
        for (int j = 0; j < 3; j++) {
            cout << "Elemento [" << i << "][" << j << "]: " << matriz[i][j];
        }
        cout << "\n"; // Quebra de linha após cada linha da matriz
    }

    return 0;
}
```

# Ponteiros, funções e seus tipos 

## Ponteiros

Um ponteiro é uma variável que armazena o **endereço de memória de outra variável**. A principal diferença de um ponteiro para uma variável normal é que o ponteiro armazena somente a posição da variável na memória, já a variável está em um endereço de memória onde o valor é armazenado. Por exemplo, podemos declarar um ponteiro para um inteiro da seguinte forma:

```cpp
int a = 10; // Variável inteira
int *p = &a; // Ponteiro p armazena o endereço de memória da variável a

printf("Valor de a: %d\n", a); // Imprimindo o valor de a
printf("Endereço de a: %p\n", (void*)&a); // Imprimindo o endereço de memória de a
printf("Valor apontado por p: %d\n", *p); // Imprimindo o valor apontado pelo ponteiro p    
```
O endereço de memória de uma variável pode ser obtido utilizando o operador `&`, enquanto o valor armazenado no endereço apontado por um ponteiro pode ser acessado utilizando o operador `*`.

## Funções

As funções são blocos de código que realizam uma tarefa específica e podem ser chamados em diferentes partes do programa. Em C++, utilizamos as funções para executar determinado comando (bloco de código) várias vezes, sem precisar reescrever o mesmo código. A sintaxe base de uma função é a seguinte:


```cpp
tipo_retorno nome_funcao(parametros) {
    // Corpo da função
}
```

As funções permitem a reutilização de código e a organização do programa em módulos menores.


### Funções sem retorno (void)

Funções sem retorno, também conhecidas como funções `void`, são aquelas que não retornam nenhum valor ao serem chamadas. Por exemplo, vamos supor que precisamos criar um separador `"------"` para dividir diferentes seções do programa, ao invés de escrever o mesmo código várias vezes, podemos criar uma função que realiza essa tarefa:

```cpp
#include <iostream>
using namespace std;

void imprimirSeparador() {
    cout << "------ \n"; // Imprime o separador
}

int main() {
    imprimirSeparador(); // Chama a função para imprimir o separador
    return 0;
}
```
Toda vez que precisarmos imprimir o separador, podemos simplesmente chamar a função `imprimirSeparador()`, evitando a repetição de código.



### Funções Inteiras (int)

Funções inteiras são aquelas que retornam um valor do tipo `int`. Por exemplo, podemos criar uma função que calcula a soma de dois números inteiros e retorna o resultado:

```cpp
int somar(int a, int b) {
    return a + b; // Retorna a soma de a e b
}
int main() {
    int resultado = somar(5, 3); // Chama a função somar
    cout << "Resultado da soma: " << resultado; // Imprime o resultado
    return 0;
}
```

Também podemos passar o retorno de uma função para uma variável, como no exemplo acima, onde o resultado da função `somar` é armazenado na variável `resultado`.

```cpp
resultado = somar(10, 20); // Chama a função somar e armazena o resultado
printf("Resultado da soma: %d\n", resultado); // Imprime o resultado
```

A ordem das funções não importa, desde que a função seja declarada antes de ser chamada. Por exemplo, podemos declarar a função `somar` antes do `main`, ou podemos declarar um protótipo da função antes do `main` e definir a função depois.

```cpp
#include <iostream>
using namespace std;

// Protótipo da função
int somar(int a, int b); // Declaração da função somar

int main() {
    int resultado = somar(5, 3); // Chama a função somar
    cout << "Resultado da soma: " << resultado; // Imprime o resultado
    return 0;
}

// Definição da função
int somar(int a, int b) {
    return a + b; // Retorna a soma de a e b
}
```


### Funções de caracteres (char)

Funções de caracteres são aquelas que retornam um valor do tipo `char`. Por exemplo, podemos criar uma função que retorna a primeira letra de uma string:

```cpp
string palavra(string str) {
    return str;  // Retorna a string recebida como parâmetro
}

int main() {
    string nome = "Alice"; // Declaração de uma string
    string palavraRetornada = palavra(nome); // Chama a função palavra
    cout << "Palavra retornada: " << palavraRetornada; // Imprime a palavra retornada
    return 0;
}
```
Observe que a função `palavra` recebe uma string como parâmetro e retorna a mesma string. No exemplo acima, chamamos a função `palavra` passando a variável `nome` como argumento, e o valor retornado é armazenado na variável `palavraRetornada`, que é então impressa no console.


### Funções booleans (bool)

Funções booleanas são aquelas que retornam um valor do tipo `bool`, que pode ser `true` ou `false`. Para utilizarmos essas funções precisamos incluir a biblioteca `<stdbool.h>` em C ou utilizar o tipo `bool` diretamente em C++. Por exemplo, podemos criar uma função que verifica se um número é par:

```cpp
#include <iostream>
#include <stdbool.h> // Biblioteca para utilizar o tipo bool em C
using namespace std;

bool retornaBooleano(){
    return true; // Retorna true
}

int main() {
    bool resultado = retornaBooleano(); // Chama a função retornaBooleano
    
    if (resultado) {
        cout << "O resultado é verdadeiro"; // Imprime se o resultado for verdadeiro
    } else {
        cout << "O resultado é falso"; // Imprime se o resultado for falso
    }
}
```


### Função para limpar a tela 

Em C++, existe uma função que permite limpar a tela do console, mas ela não é padronizada e pode variar dependendo do sistema operacional. No Windows, podemos utilizar o comando `system("cls")`, enquanto no Linux e macOS, podemos utilizar `system("clear")`. A seguir, um exemplo de como criar uma função para limpar a tela:

```cpp
#include <iostream>
#include <cstdlib> // Biblioteca para utilizar a função system
using namespace std;

int a; 

printf("Digite um número: ");
scanf("%d", &a); // Lê um número do usuário

system("cls"); // Limpa a tela (Windows)
``` 

Para não precisar escrever o comando `system("cls")` ou `system("clear")` várias vezes, podemos criar uma função que realiza essa tarefa:

```cpp
#include <iostream>
#include <cstdlib> // Biblioteca para utilizar a função system
using namespace std;

void limparTela() {
    system("cls"); // Limpa a tela no Windows

}

int main() {
    int a; 

    printf("Digite um número: ");
    scanf("%d", &a); // Lê um número do usuário

    limparTela(); // Chama a função para limpar a tela

    return 0;
}
```

# Funções com Parâmetros 

Funções com parâmetros são aquelas que recebem valores de entrada para realizar operações específicas. Os parâmetros permitem que as funções sejam mais flexíveis e reutilizáveis, pois podem operar com diferentes valores fornecidos pelo usuário ou por outras partes do programa.

Sua sintaxe básica é a seguinte:

```cpp
tipo_retorno nome_funcao(tipo_parametro1 parametro1, tipo_parametro2 parametro2, ...) {
    // Corpo da função
}
```

## Funções com parâmetros unicos 

Funções com parâmetros únicos são aquelas que recebem apenas um valor de entrada. Por exemplo, podemos criar uma função que calcula o quadrado de um número inteiro:

```cpp
void imprimirQuadrado(int numero) {
    int quadrado = numero * numero; // Calcula o quadrado do número
    printf("O quadrado de %d é: %d\n", numero, quadrado); // Imprime o resultado
}

int main() {
    int valor;
    printf("Digite um número: ");
    scanf("%d", &valor); // Lê um número do usuário
    imprimirQuadrado(valor); // Chama a função com o valor fornecido pelo usuário

    return 0;
}
``` 


## Funções com múltiplos parâmetros

Funções com múltiplos parâmetros são aquelas que recebem mais de um valor de entrada. Por exemplo, podemos criar uma função que calcula a soma de dois números inteiros:

```cpp
int somar(int a, int b) {
    printf("A soma de %d e %d é: %d\n", a, b, a + b); // Imprime o resultado da soma
} 

int main(){
    int num1 = 5, num2 = 10;
    somar(num1, num2); // Chama a função com dois parâmetros
}
```

## Funções com ponteiros e referências como parâmetros

Funções com ponteiros e referências como parâmetros permitem que os valores das variáveis sejam modificados diretamente dentro da função, sem a necessidade de retornar um valor. Sua sintaxe básica é a seguinte:

```cpp
tipo_retorno nome_funcao(tipo_parametro1* parametro1, tipo_parametro2& parametro2, ...) {
    // Corpo da função
}
```
O ponteiro é indicado pelo operador `*`, enquanto a referência é indicada pelo operador `&`. Por exemplo, `int *num` indica que `num` é um ponteiro que aponta para um inteiro.

Em C++, podemos utilizar ponteiros e referências para passar variáveis para funções, sem a necessidade de fazer atribuição, visto que os valores podem ser alterados diretamente. 

Um exemplo de função que utiliza ponteiros para modificar o valor de uma variável é mostrado abaixo:
```cpp
//Sem ponteiros
int retornaDobro(int num){
    return num * 2; // Retorna o dobro do número fornecido
}

int main() {
    int valor = 5;
    int resultado = retornaDobro(valor); // Chama a função com o valor fornecido
    printf("O dobro de %d é: %d\n", valor, resultado); 
    return 0;
}
```

```cpp
//Com ponteiros
void dobro(int *num) {
    *num = *num * 2; // Modifica o valor do número diretamente
}

int main() {
    int valor = 5;
    dobro(&valor); // Chama a função com o endereço do valor
    printf("O dobro de %d é: %d\n", valor, valor);
    return 0;
}
```
Observe que na função `dobro`, utilizamos um ponteiro para modificar o valor da variável `valor` diretamente, sem a necessidade de retornar um valor. A função recebe o endereço da variável como parâmetro e altera seu conteúdo utilizando o operador `*`.



## Funções com vetores como parâmetros

Funções com vetores como parâmetros permitem que um vetor seja passado para uma função, possibilitando a manipulação de seus elementos dentro da função. Para fazer isso precisamos passar o vetor como um ponteiro, juntamente com seu tamanho, para que a função saiba quantos elementos existem no vetor.

Um exemplo de função que recebe um vetor como parâmetro e imprime seus elementos é mostrado abaixo:

```cpp
void imprimirVetor(int *vetor, int tamanho) {
    for (int i = 0; i < tamanho; i++) {
        printf("Elemento %d: %d\n", i, vetor[i]); // Imprime cada elemento do vetor
    }
}

int main() {
    int numeros[] = {10, 20, 30, 40, 50}; 
    imprimirVetor(numeros, 5);
    return 0;
}
```

Como usamos ponteiros, não precisamos retornar valores caso seja necessário alterar os valores do vetor, pois podemos modificar diretamente os elementos do vetor dentro da função. Por exemplo, podemos criar uma função que dobra os valores de um vetor:

```cpp
void alterarVetor(int *vetor, int tamanho) {
    for (int i = 0; i < tamanho; i++) {
        vetor[i] = vetor[i] +1; // Adiciona 1 a cada elemento do vetor
    }
}

int main() {
    int numeros[] = {10, 20, 30, 40, 50}; 
    alterarVetor(numeros, 5); 
    imprimirVetor(numeros, 5); 
    return 0;
}
```

# Alocação dinâmica de memória

A alocação dinâmica de memória é um recurso que permite ao programa solicitar memória durante a execução, em vez de depender de uma quantidade fixa de memória definida em tempo de compilação. 


## Incluindo arquivos em outros 

A estrutura de um programa em C++ pode ser organizada em múltiplos arquivos, permitindo a separação de funcionalidades e a reutilização de código. Para incluir arquivos em outros, utilizamos a diretiva `#include`, que permite incorporar o conteúdo de um arquivo em outro. 

Por padrão, chamamos o arquivo principal de `main.cpp`, e os outros arquivos que contem funções auxiliares podem ser chamados de `funcoes.cpp`, `funcoes.h`, entre outros. A extensão `.h` é utilizada para arquivos de cabeçalho, que contêm declarações de funções, classes e variáveis globais, enquanto a extensão `.cpp` é utilizada para arquivos de implementação, que contêm o código das funções.

Para incluir um arquivo em outro, supondo que temos um arquivo `funcoes.h`, que contenha a função `imprimirOla`, utilizamos a seguinte sintaxe:

```cpp
#include "funcoes.h" // Inclui o arquivo de cabeçalho funcoes.h

int main() {
    imprimirOla(); // Chama a função definida em funcoes.h
    return 0;
}
```


## Alocação dinâmica de vetores (C)

A alocação dinâmica de vetores permite criar vetores cujo tamanho é determinado em tempo de execução, ao invés de ser fixo em tempo de compilação. Em C, podemos utilizar a função `malloc` para alocar memória dinamicamente para um vetor, e a função `free` para liberar a memória alocada quando não for mais necessária.

Para alocar dinamicamente um vetor, vamos criar uma função `alocaVetor` que recebe o tamanho do vetor como parâmetro e retorna um ponteiro para o vetor alocado. A função utiliza o operador `malloc` para alocar memória para o vetor e retorna o ponteiro para o vetor alocado.

```cpp
int* alocaVetor(int tamanho) {
    int *aux; 
    aux = (int*) malloc(tamanho * sizeof(int)); // Aloca memória para o vetor
    return aux; 
}

int main(){
    int *vetor, tamanho;
    printf("Digite o tamanho do vetor: ");
    scanf("%d", &tamanho); 
    vetor = alocaVetor(tamanho); // Aloca memória para o vetor

    // Aqui você pode usar o vetor alocado normalmente, como atribuir valores e imprimir os elementos

    free(vetor); // Libera a memória alocada
}
```
Analisando a função `alocaVetor`, podemos observar que ela recebe um parâmetro `tamanho`, que indica o número de elementos que o vetor deve ter. Em seguida, a função utiliza o operador `malloc` para alocar memória suficiente para armazenar `tamanho` elementos do tipo `int`. O `(int*)` é um cast que converte o ponteiro retornado por `malloc` para o tipo `int*`, garantindo que o ponteiro seja do tipo correto. Por fim, a função retorna o ponteiro para o vetor alocado, permitindo que ele seja utilizado no programa principal.


## Alocação dinâmica de vetores (C++)

Em C++, a alocação dinâmica de vetores pode ser feita utilizando o operador `new`, para isso, precisamos incluir a biblioteca `<new>`, que fornece suporte para alocação dinâmica de memória. A seguir, um exemplo de como alocar e liberar memória para um vetor em C++:

Aqui está um exemplo de como alocar e liberar memória para um vetor em C++:

```cpp
#include <new>

int main(){
    int tamanho;
    printf("Digite o tamanho do vetor: ");
    scanf("%d", &tamanho);

    int *vetor = new int[tamanho]; // Aloca memória para o vetor

    // Aqui você pode usar o vetor alocado normalmente, como atribuir valores e imprimir os elementos

    free(vetor); // Libera a memória alocada
    return 0;
}
``` 

## Alocação dinâmica de matrizes (C)

Similarmente à alocação dinâmica de vetores, podemos alocar dinamicamente uma matriz em C utilizando ponteiros e a função `malloc`. Mas neste caso, vamos utilizar um **ponteiro para ponteiro**, ou seja, um ponteiro que aponta para um vetor de ponteiros que representa as linhas, e cada um dos ponteiros aponta para um vetor de elementos que representa uma coluna da matriz, permitindo criar uma estrutura bidimensional. A seguir, um exemplo de como alocar e liberar memória para uma matriz em C:

```cpp
int main(){
    int linhas = 3, colunas = 3;
    int **matriz;

    //Alocando memória para as linhas da matriz
    matriz = (int**) malloc(linhas * sizeof(int*)); 

    //Alocando memória para as colunas de cada linha da matriz
    for (int i = 0; i < linhas; i++) {
        matriz[i] = (int*) malloc(colunas * sizeof(int)); 
    }

    // Aqui você pode usar a matriz alocada normalmente, como atribuir valores e imprimir os elementos
    for (int i = 0; i < linhas; i++) {
        for (int j = 0; j < colunas; j++) {
            matriz[i][j] = i; 
            printf("%d ", matriz[i][j]);
        }
        }
        printf("\n");
    }

    free(matriz); // Libera a memória alocada para a matriz
}
```


## Alocação dinâmica de matrizes (C++)

Em C++, a alocação dinâmica de matrizes pode ser feita de maneira similar à alocação dinâmica de vetores, utilizando o operador `new`. A seguir, um exemplo de como alocar e liberar memória para uma matriz em C++:

```cpp
#include <new>

int main(){
    int linhas = 3, colunas = 3;
    int **matriz;

    //Alocando memória para as linhas da matriz
    matriz = new int*[linhas]; 

    //Alocando memória para as colunas de cada linha da matriz
    for (int i = 0; i < linhas; i++) {
        matriz[i] = new int[colunas]; 
    }

    // Aqui você pode usar a matriz alocada normalmente, como atribuir valores e imprimir os elementos
}
```

# Structs 

## O que são Structs?

Uma `struct` (estrutura) é um tipo de dado definido pelo usuário que permite agrupar variáveis de diferentes tipos sob um único nome, facilitando a organização e o gerenciamento de dados relacionados. 

Sua sintaxe básica é a seguinte:

```cpp
struct NomeDaStruct {
    tipo_variavel1 nome_variavel1;
    tipo_variavel2 nome_variavel2;
    // ... outras variáveis
};
```
Vamos criar uma `struct` chamada `Palavra` que contém uma variável do tipo `string` chamada `palavra` e uma variável do tipo `int` chamada `tamanho`. A seguir, um exemplo de como declarar e utilizar a `struct`:

```cpp
#include <iostream>
#include <string>

struct Palavra {
    int tamanho; // Variável do tipo int
    char letra; // Variável do tipo char
    char palavra[255]; // Variável do tipo string
    
};

int main() {
    
    // Criando uma variável do tipo Palavra
    struct Palavra minhaPalavra;

    // Modificando os valores das variáveis da struct
    minhaPalavra.tamanho = 5; // Atribuindo valor à variável tamanho
    minhaPalavra.letra = 'A'; // Atribuindo valor à variável letra
    strcpy(minhaPalavra.palavra, "Exemplo"); // Atribuindo valor à variável palavra

    // Imprimindo os valores das variáveis da struct
    printf("Tamanho: %d\n", minhaPalavra.tamanho);
    printf("Letra: %c\n", minhaPalavra.letra);
    printf("Palavra: %s\n", minhaPalavra.palavra);
}
```
Observe que para copiar um string para a variável `palavra` da struct, utilizamos a função `strcpy` da biblioteca `<cstring>`, que permite copiar o conteúdo de uma string para outra. Além disso, podemos acessar os membros da struct utilizando o operador `.` (ponto), como mostrado no exemplo acima.


### Lista de Structs

Para criar uma lista de structs, podemos utilizar um vetor de structs, onde cada elemento do vetor representa uma instância da struct. A seguir, um exemplo de como criar e utilizar uma lista de structs:

```cpp
#include <iostream>
#include <string>

struct Palavra {
    int tamanho;
    char letra;
    char palavra[255];
};

int main() {
    // Criando uma lista de structs
    struct Palavra listaPalavras[3];

    // Preenchendo os valores da lista
    listaPalavras[0].tamanho = 5;
    listaPalavras[0].letra = 'A';
    strcpy(listaPalavras[0].palavra, "Exemplo");

    listaPalavras[1].tamanho = 7;
    listaPalavras[1].letra = 'B';
    strcpy(listaPalavras[1].palavra, "Outro");

    listaPalavras[2].tamanho = 9;
    listaPalavras[2].letra = 'C';
    strcpy(listaPalavras[2].palavra, "Terceiro");

    // Imprimindo os valores da lista
    for (int i = 0; i < 3; i++) {
        printf("Palavra %d: %s, Tamanho: %d, Letra: %c\n", i + 1, listaPalavras[i].palavra, listaPalavras[i].tamanho, listaPalavras[i].letra);
    }

    return 0;
}
```

## Estruturas com sub-estruturas (Structs aninhadas)

Uma struct pode conter outras structs como membros, permitindo a criação de estruturas mais complexas e organizadas. Isso é conhecido como structs aninhadas. 

Sintaxe básica de uma struct aninhada:

```cpp
struct NomeDaStruct {
    tipo_variavel1 nome_variavel1;
    struct OutraStruct {
        tipo_variavel2 nome_variavel2;
        // ... outras variáveis
    } nome_outra_struct;
};
```

Um exemplo de uso, vamos criar a struct `Aluno`, que contém uma sub-struct `Data` para armazenar a data de nascimento do aluno. A seguir, um exemplo de como declarar e utilizar structs aninhadas:

```cpp
#include <stdio.h>

typedef struct Data{
    int dia;
    int mes;
    int ano;
} Data;

struct Aluno{
    int id;
    Data dataNascimento;
};

int main() {
    // Criando uma variável do tipo Aluno
    struct Aluno aluno1;

    // Modificando os valores das variáveis da struct
    aluno1.id = 1; // Atribuindo valor à variável id
    aluno1.dataNascimento.dia = 15; // Atribuindo valor à variável dia da sub-struct Data
    aluno1.dataNascimento.mes = 6; // Atribuindo valor à variável mes da sub-struct Data
    aluno1.dataNascimento.ano = 2000; // Atribuindo valor à variável ano da sub-struct Data

    // Imprimindo os valores das variáveis da struct
    printf("ID: %d\n", aluno1.id);
    printf("Data de Nascimento: %02d/%02d/%04d\n", aluno1.dataNascimento.dia, aluno1.dataNascimento.mes, aluno1.dataNascimento.ano);

    return 0;
}
```

## Structs em C++

De forma similar ao C, em C++ podemos utilizar structs para agrupar variáveis de diferentes tipos sob um único nome. A sintaxe básica é a mesma, mas em C++ podemos utilizar recursos adicionais, como construtores e métodos dentro das structs.

Sintaxe básica de uma struct em C++:


Vamos criar uma struct chamada `Fruta`, que contém uma variável do tipo `string` chamada `cor` e uma variável do tipo string chamada `nome`. A seguir, um exemplo de como declarar e utilizar a struct em C++:

```cpp
#include <stdio.h>
#include <iostream>
#include <string>
#include <new>

using namespace std;

struct Fruta {
    string cor; // Variável do tipo string
    string nome; // Variável do tipo string
};

int main(){
    // Criando uma variável do tipo Fruta
    fruta *primeiraFruta = new Fruta; // Alocando memória dinamicamente para a struct

    // Modificando os valores das variáveis da struct
    primeiraFruta->cor = "Vermelha"; // Atribuindo valor à variável cor
    primeiraFruta->nome = "Maçã"; // Atribuindo valor à variável nome

    // Imprimindo os valores das variáveis da struct
    cout << "Cor: " << primeiraFruta->cor << endl;
    cout << "Nome: " << primeiraFruta->nome << endl;

    return 0;
}
```

### Criando uma lista de structs em C++

Para criar uma lista de structs em C++, podemos utilizar um vetor de structs, onde cada elemento do vetor representa uma instância da struct. A seguir, um exemplo de como criar e utilizar uma lista de structs em C++:

```cpp
#include <iostream>
#include <string>

using namespace std;

struct Fruta {
    string cor;
    string nome;
};

int main() {
    // Criando um vetor de structs
    Fruta listaFrutas[3];

    // Preenchendo os valores das structs
    listaFrutas[0].cor = "Vermelha";
    listaFrutas[0].nome = "Maçã";

    listaFrutas[1].cor = "Verde";
    listaFrutas[1].nome = "Uva";

    listaFrutas[2].cor = "Amarela";
    listaFrutas[2].nome = "Banana";

    // Imprimindo os valores das structs
    for (int i = 0; i < 3; i++) {
        cout << "Cor: " << listaFrutas[i].cor << ", Nome: " << listaFrutas[i].nome << endl;
    }

    return 0;
}
```

# Leitura e escrita em arquivos externos 


## Escrever em arquivos

Para abrir e escrever em um arquivo em C++, podemos utilizar a biblioteca `<fstream>`, que fornece classes para manipulação de arquivos. Caso o arquivo não exista, ele será criado automaticamente, muito comum para escrever arquivos de log, por exemplo. A seguir, um exemplo de como abrir e escrever em um arquivo em C++:

```cpp
#include <iostream>
#include <fstream> // Biblioteca para manipulação de arquivos

using namespace std;

int main(){
    // Função para criar o ojeto de arquivo e escrever nele
    ofstream arquivoDeSaida;

    // Abrindo o arquivo para escrita
    arquivoDeSaida.open("arquivo.txt", std::ios_base::app); // Abre o arquivo "arquivo.txt"

    // Escrevendo no arquivo
    arquivoDeSaida << "Olá, mundo!"; // Escreve "Olá, mundo!" no arquivo

    // Fechando o arquivo
    arquivoDeSaida.close(); // Fecha o arquivo
}
```

## Ler conteúdo de arquivos

Para abrir e ler o conteúdo de um arquivo em C, é necessário criar um ponteiro para o arquivo utilizando a função `fopen`, que recebe como parâmetros o nome do arquivo e o modo de abertura (leitura, escrita, etc.). Em seguida, podemos utilizar a função `fscanf` para ler os dados do arquivo e armazená-los em variáveis. Por fim, é importante fechar o arquivo utilizando a função `fclose`. A seguir, um exemplo de como abrir e ler o conteúdo de um arquivo em C: 

```cpp
#include <stdio.h>

int main() {
    FILE *arquivo; // Ponteiro para o arquivo

    arquivo = fopen("arquivo.txt", "r"); // Abre o arquivo "arquivo.txt" para leitura

    if (arquivo) {
        // Laço para ler o conteúdo do arquivo até o final
        while((int c = getc(arquivo)) != EOF) {
            printf("%c", c); // Imprime cada caractere do arquivo
        }
        fclose(arquivo); // Fecha o arquivo
    }
    return 0;
}
```

### Ler conteúdo de arquivos em C++

Para abrir e ler o conteúdo de um arquivo em C++, podemos utilizar a biblioteca `<fstream>`, que fornece classes para manipulação de arquivos. A seguir, um exemplo de como abrir e ler o conteúdo de um arquivo em C++:

```cpp
#include <iostream>
#include <fstream> // Biblioteca para manipulação de arquivos

using namespace std;

int main() {
    
    ifstream input("arquivo.txt"); // Abre o arquivo "arquivo.txt" para leitura

    string textoLido; // Variável para armazenar o conteúdo lido do arquivo

    for (string linha; getline(input, linha); ) { // Laço para ler cada linha do arquivo
        textoLido += linha; // Adiciona a linha lida à variável textoLido
    }

    cout << textoLido; // Imprime o conteúdo lido do arquivo
```

## Limpando o conteúdo de um arquivo

Para limpar o conteúdo de um arquivo em C++, podemos abrir o arquivo em modo de escrita, sobrescrevendo o conteúdo existente usando `""`. A seguir, um exemplo de como limpar o conteúdo de um arquivo em C++:

```cpp
#include <iostream>
#include <fstream> // Biblioteca para manipulação de arquivos

using namespace std;

int main() {
    ofstream arquivoDeSaida;

    // Abrindo o arquivo para escrita, sobrescrevendo o conteúdo existente
    arquivoDeSaida.open("arquivo.txt");

    arquivoDeSaida << ""; // Escreve uma string vazia no arquivo, limpando seu conteúdo

    // Fechando o arquivo
    arquivoDeSaida.close(); // Fecha o arquivo

    return 0;
}
```


# Convertendo código C para C++

Vamos aprender sobre alguns comandos que podem ser utilizados para converter código C para C++.

## `cout` e `cin`

O comando `printf` é utilizado em C para imprimir informações no console, enquanto o comando `cout` é utilizado em C++ para a mesma finalidade. A principal diferença entre eles é que `cout` faz parte da biblioteca padrão de C++ e utiliza o operador de inserção `<<` para enviar dados para a saída padrão. 

Já o comando `scanf` é utilizado em C para ler informações do console, enquanto o comando `cin` é utilizado em C++ para a mesma finalidade. A principal diferença entre eles é que `cin` faz parte da biblioteca padrão de C++ e utiliza o operador de extração `>>` para receber dados da entrada padrão.

Um exemplo de como utilizar `cout` e `cin` em C++ é mostrado abaixo:

```cpp
#include <iostream> // Biblioteca para entrada e saída de dados

int main() {
    int idade; // Variável inteira
    double salario; // Variável double
    char letra; // Variável char
    bool ativo; // Variável bool
    string nome; // Variável string

    // Solicitando ao usuário que digite os valores das variáveis
    std::cout << "Digite a idade: ";
    std::cin >> idade;
    std::cout << "Digite o salário: ";
    std::cin >> salario;
    std::cout << "Digite a letra: ";
    std::cin >> letra;
    std::cout << "Digite se está ativo (1 para sim, 0 para não): ";
    std::cin >> ativo;
    std::cout << "Digite o nome: ";
    std::cin >> nome;

    // Imprimindo os valores das variáveis utilizando cout
    std::cout << "Idade: " << idade << std::endl;
    std::cout << "Salário: " << salario << std::endl;
    std::cout << "Letra: " << letra << std::endl;
    std::cout << "Ativo: " << (ativo ? "Sim" : "Não") << std::endl;
    std::cout << "Nome: " << nome << std::endl;

    return 0;
}
```
Para evitar repetir o `std::` antes de cada comando, podemos utilizar a diretiva `using namespace std;`, que permite utilizar os comandos da biblioteca padrão de C++ sem precisar especificar o namespace. A seguir, um exemplo de como utilizar `cout` em C++ com a diretiva `using namespace std;`:

```cpp
#include <iostream> // Biblioteca para entrada e saída de dados
#include <string> // Biblioteca para manipulação de string  

using namespace std; // Permite utilizar os comandos da biblioteca padrão sem especificar o namespace

int main() {
    int idade; // Variável inteira
    double salario; // Variável double
    char letra; // Variável char
    bool ativo; // Variável bool
    string nome; // Variável string

    // Solicitando ao usuário que digite os valores das variáveis
    cout << "Digite a idade: ";
    cin >> idade;
    cout << "Digite o salário: ";
    cin >> salario;
    cout << "Digite a letra: ";
    cin >> letra;
    cout << "Digite se está ativo (1 para sim, 0 para não): ";
    cin >> ativo;
    cout << "Digite o nome: ";
    cin >> nome;   

    // Imprimindo os valores das variáveis utilizando cout
    cout << "Idade: " << idade << endl;
    cout << "Salário: " << salario << endl;
    cout << "Letra: " << letra << endl;
    cout << "Ativo: " << (ativo ? "Sim" : "Não") << endl;
    cout << "Nome: " << nome << endl;

    return 0;
}
```

## Usando Strings 

Em C, para manipular as strings, é necessário utilizar a biblioteca `<string.h>`, que fornece funções para manipulação de strings, como `strcpy`, `strlen`, `strcmp`, entre outras. Já em C++, podemos utilizar a biblioteca `<string>`, que fornece a classe `std::string`, que facilita a manipulação de strings, permitindo operações como concatenação, comparação, busca, entre outras. 

Um exemplo de como utilizar strings em C++ é mostrado abaixo:

```cpp
#include <iostream> // Biblioteca para entrada e saída de dados
#include <string> // Biblioteca para manipulação de strings

using namespace std; // Permite utilizar os comandos da biblioteca padrão sem especificar o namespace

int main() {
    string nome = "Alice"; // Declaração de uma string
    string sobrenome = "Silva"; // Declaração de outra string

    // Concatenando as strings
    string nomeCompleto = nome + " " + sobrenome; // Concatenação das strings utilizando o operador +  

    // Imprimindo o nome completo
    cout << "Nome completo: " << nomeCompleto << endl; // Imprime o nome completo no console
    
    return 0;
}
```
